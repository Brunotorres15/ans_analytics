```python
%sql

CREATE OR REPLACE TABLE dev.gold.DIM_PORTE_OPERADORA
LOCATION "s3://novadrive-project/gold/dim_porte_operadora"
AS
SELECT 
  monotonically_increasing_id() AS SK_PORTE_OPERADORA,
  PORTE_OPERADORA AS NM_PORTE_OPERADORA
FROM (
  SELECT DISTINCT PORTE_OPERADORA FROM dev.silver.silver_sip
);
```
