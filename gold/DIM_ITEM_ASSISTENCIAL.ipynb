```python
%sql

CREATE OR REPLACE TABLE dev.gold.DIM_ITEM_ASSISTENCIAL
LOCATION "s3://novadrive-project/gold/dim_item_assistencial"
AS
SELECT 
  monotonically_increasing_id() AS SK_ITEM_ASSISTENCIAL,
  ID_ITEM_ASST,
  DE_ITEM_ASST AS NM_ITEM_ASSISTENCIAL
FROM (
  SELECT DISTINCT ID_ITEM_ASST, DE_ITEM_ASST FROM dev.silver.silver_sip
);
```
